# Análisis de Calidad de Datos
Este notebook muestra el procedimiento de Exploratory Data Analysis y la limpieza inicial del dataset `ai_dependency_career_anxiety_students.csv`.


```python
import pandas as pd
from pathlib import Path

# Rutas
project_root = Path('..')
raw_path = project_root / 'data' / 'raw' / 'ai_dependency_career_anxiety_students.csv'

df = pd.read_csv(raw_path)
df.head()
```

## 1. Información básica del dataset


```python
print('Filas:', df.shape[0])
print('Columnas:', df.shape[1])
print('Tipos de dato:')
print(df.dtypes)
```

## 2. Variables numéricas, categóricas y binarias


```python
numeric_cols = df.select_dtypes(include='number').columns.tolist()
categorical_cols = df.select_dtypes(exclude='number').columns.tolist()
binary_cols = [c for c in df.columns if df[c].nunique() == 2]
print('Numéricas:', numeric_cols[:5], '...')
print('Categóricas:', categorical_cols[:5], '...')
print('Binarias:', binary_cols)
```

## 3. Detección de valores nulos y duplicados


```python
null_counts = df.isnull().sum()
print('Valores nulos por columna:')
print(null_counts[null_counts > 0])
dup_rows = df[df.duplicated(keep=False)]
print('Filas duplicadas:', dup_rows.shape[0])
```

## 4. Propuesta de limpieza inicial (aplicada en `run_eda.py`)
- Normalización de nombres de columnas a snake_case.
- Eliminación de filas duplicadas exactas.
- Eliminación de filas con valores nulos (por simplicidad).
- Guardado del dataset limpio en `data/processed/dataset_limpio.csv`.
